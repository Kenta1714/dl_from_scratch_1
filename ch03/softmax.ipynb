```python
import numpy as np
```

```python
a = np.array([0.3, 2.9, 4.0])
```

```python
exp_a = np.exp(a)
exp_a
```

```python
sum_exp_a = np.sum(exp_a)
sum_exp_a
```

```python
y = exp_a / sum_exp_a
print(y)
print(np.sum(y))
```
